```python
total_respondents = df['ResponseId'].nunique()
print(f"Загальна кількість респондентів: {total_respondents}")
```

2

```python
questions = set(df_schema['qname'].unique()) & set(df.columns)

df_filtered_questions = df[list(questions)]
respondents_complete_answers = df_filtered_questions.dropna().shape[0]

print(f"Кількість респондентів, які відповіли на всі спільні питання: {respondents_complete_answers}")
print(f"Загальна кількість респондентів: {total_respondents}")

percentage_complete = (respondents_complete_answers / total_respondents) * 100
print(f"Відсоток повноти відповідей: {percentage_complete:.2f}%")
```

3

```python
print(f'Міри центральної тенденції:')

mean_value = df['ConvertedCompYearly'].mean()
print(f'Середнє значення: {round(mean_value, 2)} $/рік')

median_value = df['ConvertedCompYearly'].median()
print(f'Медіана: {round(median_value, 2)} $/рік')

mode_value = df['ConvertedCompYearly'].mode()
print(f'Мода:{round(mode_value, 2)} $/рік')
```

4

```python
remote_workers_count = df[df['Employment'] == 'Remote'].shape[0]
print(f"Кількість респондентів, які працюють віддалено (Remote): {remote_workers_count}")
```

5

```python
df['work_with_python'] = df['LanguageHaveWorkedWith'].str.contains('Python', case=False, na=False)

python_users = df['work_with_python'].sum()

python_percentage = (python_users / total_respondents) * 100

print(f"Відсоток респондентів: {python_percentage:.2f}%")
```

6

```python
df['Learned_with_online_courses'] = df['LearnCode'].str.contains('Online Courses or Certification', case=False, na=False)

# print(df[['LearnCode', 'Learned_with_online_courses']].head())

online_course_learners = df['Learned_with_online_courses'].sum()

print(f"Кількість респондентів, які навчалися через онлайн-курси: {online_course_learners}")


```

7

```python
python_users = df[df['work_with_python'] == True]

python_users_cleaned = python_users.dropna(subset=['ConvertedCompYearly'])

py_progers_by_country = python_users_cleaned.groupby('Country')['ConvertedCompYearly'].agg([
    ('Average Compensation (USD)', 'mean'),
    ('Median Compensation (USD)', 'median')
]).reset_index()

display(py_progers_by_country.head())
```

8

```python
highest_paid_specialists = df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']]

highest_paid_specialists = highest_paid_specialists.sort_values(by='ConvertedCompYearly', ascending=False)

highest_paid_specialists = highest_paid_specialists.reset_index(drop=True)

top_5_highest_paid = highest_paid_specialists.head(5)

display(top_5_highest_paid)
```
